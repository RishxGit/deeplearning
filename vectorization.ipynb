```python
#SIGMOID FUCTION

def sigmoid(x):

    s = 1/(1 + np.exp(-x))

    return s
```

```python
sigmoid(3)
```

```python
x = np.array([1,2,3])
sigmoid(x)
```

```python
print(x+3)
```

```python
print(np.exp(x))
```

```python
#SIGMOID GRADIENT

def sigmoid_dervative(x):

    s = sigmoid(x)
    ds = s * (1 - s)

    return ds
```

```python
x = np.array([1,2,3])
print('sigmoid_dervative:', sigmoid_dervative(x))
```

```python
#softmax

def softmax(x):
    x_exp = np.exp(x)
    total = np.sum(x_exp,axis=1,keepdims=True)
    x = x_exp/total

    return x
```

```python
x = np.array([[9,2,5,0,0],
            [7,5,0,0,0]])
print("softmax:\n", softmax(x))
```

```python
#L1 LOSS FUNCTION:   
#     loss = np.sum(abs(y-yhat))

def L1(yhat,y):
    diff = yhat - y
    x = np.abs(diff)
    l1 = np.sum(x)

    return l1
```

```python
#L2 LOSS FUNCTION:
#   loss = np.dot(y-yhat,y-yhat)

def L2(yhat,y):
    diff = yhat - y
    diff_2 = diff**2
    l2 = np.sum(diff_2)

    return l2
```

```python
yhat = np.array([.9, 0.2, 0.1, .4, .9])
y = np.array([1, 0, 0, 1, 1])
print('L1 :',(L1(yhat,y)))
print('L2 :',(L2(yhat,y)))
```
